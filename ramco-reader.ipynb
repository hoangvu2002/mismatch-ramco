```python
import pandas as pd
from datetime import datetime

# Hiển thị đủ số dòng và số cột mong muốn
pd.set_option("display.max_rows", 100)   # tối đa số dòng hiển thị
pd.set_option("display.max_columns", None)  # hiện tất cả các cột
pd.set_option("display.width", None)   # không giới hạn độ rộng
```

```python
# Đường dẫn file CSV
csv_path = "RAMCO-OCT-2025.csv"
```

```python
# Đọc file
df = pd.read_csv(csv_path)
print("Kích thước ban đầu:", df.shape)
df.head()
```

```python
# Xóa các cột không cần thiết
cols_to_remove = ["#", "OU", "Employee Type"]
df = df.drop(columns=[col for col in cols_to_remove if col in df.columns])
print("Sau khi xóa cột:", df.shape)
df.head()
```

```python
# Xóa hàng đầu tiên và đưa hàng chứa ngày lên làm tên cột
df.columns = df.iloc[0]   # gán hàng đầu tiên làm tên cột
df = df[1:].reset_index(drop=True)  # xóa hàng đầu tiên, reset lại index
print("Sau khi chỉnh lại header:", df.shape)
df.head()
```

```python
# Đổi tên và định dạng lại cột đầu tiên
# Giả định cột đầu tiên hiện không có tên rõ ràng (ví dụ NaN hoặc trống)
df = df.rename(columns={df.columns[0]: "Oracle_ID", df.columns[1]: "Employee Name"})
df.head()
```

```python
# Chuyển Oracle_ID về dạng string (để không bị float hoặc mất số 0 đầu)
df["Oracle_ID"] = df["Oracle_ID"].astype(str).str.strip().str.replace(r"\.0$", "", regex=True)

df.head()
```

```python
from dateutil import parser
import re

def normalize_col_name_to_mdY(col):
    # Nếu đã là datetime-like (pandas Timestamp)
    try:
        if hasattr(col, "year") and hasattr(col, "month") and hasattr(col, "day"):
            return f"{col.month}/{col.day}/{col.year}"
    except Exception:
        pass

    s = str(col).strip()

    # Thay mọi loại dash/non-breaking hyphen thành dấu '-' chuẩn
    s = re.sub(r"[\u2010\u2011\u2012\u2013\u2014\u2212]", "-", s)

    # Loại bỏ ký tự không in được (zero-width, nbsp, ...)
    s = re.sub(r"[\u00A0\u200B\u200C\u200D\uFEFF]", "", s)

    # Nếu đã là dạng số miễn là không chứa chữ, giữ nguyên (ví dụ 'Oracle_ID' hoặc '123')
    if re.search(r"[A-Za-z]", s) is None and re.search(r"\d", s) is not None:
        return s

    # Thử parse bằng định dạng cố định dd-MMM-YYYY (ví dụ 01-Oct-2025)
    for fmt in ("%d-%b-%Y", "%d-%b-%y", "%d/%b/%Y", "%d/%b/%y", "%d-%B-%Y", "%d %b %Y"):
        try:
            dt = pd.to_datetime(s, format=fmt, errors="raise")
            return f"{dt.month}/{dt.day}/{dt.year}"
        except Exception:
            pass

    # Thử pd.to_datetime với dayfirst (gợi ý: dd-MMM-YYYY)
    try:
        dt = pd.to_datetime(s, dayfirst=True, errors="coerce")
        if not pd.isna(dt):
            return f"{dt.month}/{dt.day}/{dt.year}"
    except Exception:
        pass

    # Cuối cùng thử dateutil.parser (linh hoạt hơn)
    try:
        dt = parser.parse(s, dayfirst=True, fuzzy=True)
        return f"{dt.month}/{dt.day}/{dt.year}"
    except Exception:
        pass

    # Nếu tất cả thất bại, trả về tên nguyên gốc (sau strip)
    return s

# Áp dụng cho tất cả tên cột
new_columns = [normalize_col_name_to_mdY(c) for c in df.columns]
df.columns = new_columns

print("Tên cột sau khi chuẩn hóa:")
print(df.columns.tolist())
df.tail()
```

```python
import os

# 1. Kiểm tra index có cần giữ không
if df.index.equals(pd.RangeIndex(start=0, stop=len(df), step=1)):
    print("Index chỉ là mặc định, sẽ được loại bỏ.")
    df = df.reset_index(drop=True)
else:
    print("Index chứa thông tin, sẽ được giữ lại.")

df.head()
```

```python
# 2. Đặt tên file xuất
csv_path = "ramco_processed_oct_2025.csv"
xlsx_path = "ramco_processed_oct_2025.xlsx"

# 3. Ghi file, ghi đè nếu tồn tại
df.to_csv(csv_path, index=False) #file này là để đọc trong notebook khác
df.to_excel(xlsx_path, index=False) #file này là để gửi cho bên Ops hoặc TQA hoặc đồng nghiệp

print(f"Đã ghi đè file:\n - {os.path.abspath(csv_path)}\n - {os.path.abspath(xlsx_path)}")
```
