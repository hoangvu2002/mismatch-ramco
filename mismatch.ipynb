```python
# Đọc sheet ATD_Final
atd_final_df = pd.read_excel(excel_path, sheet_name="ATD_Final")

# Đảm bảo cột Date trong atd_final_df có định dạng tháng/ngày/năm
atd_final_df['Date'] = pd.to_datetime(atd_final_df['Date'], errors='coerce').dt.strftime('%#m/%#d/%Y')


# Kiểm tra vài dòng đầu
atd_final_df.head()
```

```python

# Tính cột Actual Attendance
def compute_actual_attendance(row):
    schedule = str(row['Shift']).strip().upper()
    
    # Các trạng thái không tính attendance
    invalid_status = {"TERMINATION", "OFF CHAT", "", "0", "REMOVED"}
    if schedule in invalid_status:
        return ""
    
    # Trường hợp Mentor
    if schedule == "MENTOR":
        return "PR"
    
    # Các trường hợp còn lại giữ nguyên attendance gốc
    return row['Attendance']

atd_final_df['Actual Attendance'] = atd_final_df.apply(compute_actual_attendance, axis=1)

# Kiểm tra kết quả
atd_final_df.head()
```

```python
# Đọc 3 file CSV
ramco_df = pd.read_csv("ramco_processed_oct_2025.csv")
active_df = pd.read_csv("Active.csv")
inactive_df = pd.read_csv("Inactive.csv")

# Hiển thị vài dòng đầu tiên của mỗi file
ramco_df.head()
#active_df.head()
#inactive_df.head()
```

```python

report_base_df = atd_final_df[['OracleID', 'Date', 'Employee Name', "Actual Attendance"]].merge(
    active_df[['OracleID', 'LOB', 'Supervisor Name', 'Manager/OM Name']],
    on='OracleID',
    how='left'
)

# Hiển thị kết quả mẫu
report_base_df
```

```python
# Unpivot (melt) các cột ngày trong ramco_df
# Giả sử các cột không phải ngày là các cột định danh (ID, Name, v.v.)

id_vars = ['Oracle_ID', 'Employee Name']  # thay đổi nếu bạn có thêm các cột định danh khác
value_vars = [col for col in ramco_df.columns if col not in id_vars]

ramco_unpivot_df = ramco_df.melt(
    id_vars=id_vars,
    value_vars=value_vars,
    var_name='Date',
    value_name='Ramco Attendance'
)

# Chuẩn hóa lại cột Date nếu cần (nếu cột đang là dạng chuỗi YYYY-MM-DD)
ramco_unpivot_df['Date'] = pd.to_datetime(ramco_unpivot_df['Date'], errors='coerce').dt.strftime('%#m/%#d/%Y')

# Xem vài dòng đầu
ramco_unpivot_df
```

```python
# Đảm bảo tên cột khớp trước khi merge
ramco_unpivot_df = ramco_unpivot_df.rename(columns={'Oracle_ID': 'OracleID'})

# Gộp dữ liệu từ ramco_unpivot_df vào report_base_df theo OracleID + Date
report_base_df = report_base_df.merge(
    ramco_unpivot_df[['OracleID', 'Date', 'Ramco Attendance']],
    on=['OracleID', 'Date'],
    how='left'
)

# Kiểm tra kết quả sau khi thêm cột Ramco Attendance
report_base_df
```

```python
# Đặt khoảng ngày cần hiển thị
start_date = "10/1/2025"
end_date = "10/15/2025"

# Chuyển cột Date về dạng datetime để lọc
report_base_df['Date_dt'] = pd.to_datetime(report_base_df['Date'], errors='coerce')
start_dt = pd.to_datetime(start_date)
end_dt = pd.to_datetime(end_date)

# Lọc dữ liệu theo khoảng ngày mong muốn
filtered_report_df = report_base_df[
    (report_base_df['Date_dt'] >= start_dt) & (report_base_df['Date_dt'] <= end_dt)
]

#Những agent đã bị ter sẽ bị hiện giá trị LOB, Supervisor Name là NaN, vì thế cần map những dòng mà có LOB, hoặc Supervisor Name là NaN thì
#sẽ map nó với bảng inactive

# Hiển thị kết quả
filtered_report_df[report_base_df["Employee Name"] == "NGUYEN TRAN HOANG ANH"]
```
