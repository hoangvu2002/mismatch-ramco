```python
# Loại bỏ hàng hiện đang dùng làm tên cột và đưa hàng đầu tiên của dữ liệu lên làm header mới

# Đặt hàng đầu tiên (hiện tại là dòng dữ liệu) làm tên cột mới
df_roster.columns = df_roster.iloc[0]

# Xóa hàng đầu tiên (vì đã dùng làm header)
df_roster = df_roster[1:].reset_index(drop=True)

print("Sau khi chỉnh lại header:")
print(df_roster.columns.tolist())
df_roster.head()
```

```python
# Đọc sheet ATD_Final
atd_final_df = pd.read_excel(excel_path, sheet_name="ATD_Final")

# Đảm bảo cột Date trong atd_final_df có định dạng tháng/ngày/năm
atd_final_df['Date'] = pd.to_datetime(atd_final_df['Date'], errors='coerce').dt.strftime('%#m/%#d/%Y')


# Kiểm tra vài dòng đầu
atd_final_df
```

```python

# Tính cột Actual Attendance
def compute_actual_attendance(row):
    schedule = str(row['Shift']).strip().upper()
    
    # Các trạng thái không tính attendance
    invalid_status = {"TERMINATION", "OFF CHAT", "", "0", "REMOVED"}
    if schedule in invalid_status:
        return ""
    
    # Trường hợp Mentor
    if schedule == "MENTOR":
        return "PR"
    
    # Các trường hợp còn lại giữ nguyên attendance gốc
    return row['Attendance']

atd_final_df['Actual Attendance'] = atd_final_df.apply(compute_actual_attendance, axis=1)

# Kiểm tra kết quả
atd_final_df.head()
```

```python
# Đọc 3 file CSV
ramco_df = pd.read_csv("ramco_processed_oct_2025.csv")
active_df = pd.read_csv("Active.csv")
inactive_df = pd.read_csv("Inactive.csv")

# Hiển thị vài dòng đầu tiên của mỗi file
#ramco_df.head()
#active_df.head()
inactive_df.head()
```

```python
# Merge với active_df trước
active_merge = atd_final_df[['OracleID', 'Date', 'Employee Name', 'Actual Attendance', "SUM Productive"]].merge(
    active_df[['OracleID', 'LOB', 'Supervisor Name', 'Manager/OM Name']],
    on='OracleID',
    how='left'
)

# Merge với inactive_df (và đổi tên cột Manager Name)
inactive_merge = atd_final_df[['OracleID', 'Date', 'Employee Name', 'Actual Attendance', "SUM Productive"]].merge(
    inactive_df[['OracleID', 'LOB', 'Supervisor Name', 'Manager Name']].rename(
        columns={'Manager Name': 'Manager/OM Name'}
    ),
    on='OracleID',
    how='left'
)

# Dùng combine_first để lấy giá trị từ inactive nếu active bị thiếu
report_base_df = active_merge.combine_first(inactive_merge)

# Hiển thị kết quả mẫu
report_base_df.head()
```

```python
# Unpivot (melt) các cột ngày trong ramco_df
# Giả sử các cột không phải ngày là các cột định danh (ID, Name, v.v.)

id_vars = ['Oracle_ID', 'Employee Name']  # thay đổi nếu bạn có thêm các cột định danh khác
value_vars = [col for col in ramco_df.columns if col not in id_vars]

ramco_unpivot_df = ramco_df.melt(
    id_vars=id_vars,
    value_vars=value_vars,
    var_name='Date',
    value_name='Ramco Attendance'
)

# Chuẩn hóa lại cột Date nếu cần (nếu cột đang là dạng chuỗi YYYY-MM-DD)
ramco_unpivot_df['Date'] = pd.to_datetime(ramco_unpivot_df['Date'], errors='coerce').dt.strftime('%#m/%#d/%Y')

# Xem vài dòng đầu
ramco_unpivot_df
```

```python
# Đảm bảo tên cột khớp trước khi merge
ramco_unpivot_df = ramco_unpivot_df.rename(columns={'Oracle_ID': 'OracleID'})

# Gộp dữ liệu từ ramco_unpivot_df vào report_base_df theo OracleID + Date
report_base_df = report_base_df.merge(
    ramco_unpivot_df[['OracleID', 'Date', 'Ramco Attendance']],
    on=['OracleID', 'Date'],
    how='left'
)

# Kiểm tra kết quả sau khi thêm cột Ramco Attendance
report_base_df
```

```python
# Đặt khoảng ngày cần hiển thị
start_date = "10/1/2025"
end_date = "10/22/2025"

# Chuyển cột Date về dạng datetime để lọc
report_base_df['Date_dt'] = pd.to_datetime(report_base_df['Date'], errors='coerce')
start_dt = pd.to_datetime(start_date)
end_dt = pd.to_datetime(end_date)

# Lọc dữ liệu theo khoảng ngày mong muốn
filtered_report_df = report_base_df[
    (report_base_df['Date_dt'] >= start_dt) & (report_base_df['Date_dt'] <= end_dt)
]

#Những agent đã bị ter sẽ bị hiện giá trị LOB, Supervisor Name là NaN, vì thế cần map những dòng mà có LOB, hoặc Supervisor Name là NaN thì
#sẽ map nó với bảng inactive

# Hiển thị kết quả
filtered_report_df
```

```python

filtered_report_df = (
    filtered_report_df
    .sort_values(by='Date_dt', ascending=True)
    .drop(columns=['Date_dt'])
    .reset_index(drop=True)
)

#tạo một dataframe cho NM 
notmark_df = filtered_report_df[filtered_report_df["Ramco Attendance"] == "NM"]

filtered_report_df
```

```python
#Tạo báo cáo Ops check 

ops_check_df = filtered_report_df[filtered_report_df['Ramco Attendance'].notna()].copy()

#Chỉ giữ lại những dòng mà Actual Attendance khác Ramco Attendance

filtered_report_df = filtered_report_df[
    filtered_report_df['Actual Attendance'] != filtered_report_df['Ramco Attendance']
].reset_index(drop=True)

#filter bỏ ra những dòng Ramco Attendance là NaN (những trường hợp này là đã bị ter)

filtered_report_df = filtered_report_df[
    filtered_report_df['Ramco Attendance'].notna()
].reset_index(drop=True)

#filter bỏ ra những dòng mà Ramco Attendance là NM (sẽ có một df riêng cho NM)
filtered_report_df = filtered_report_df[filtered_report_df["Ramco Attendance"] != "NM"].reset_index(drop=True)

#filter bỏ ra những dòng Actual Attendance trong list excule codes

exclude_codes = ["", "NCNS", "PR"]


# 2) Chuẩn hóa cột 'Actual Attendance' để so sánh an toàn:
#    - chuyển về string
#    - strip khoảng trắng
#    - đưa NaN về chuỗi rỗng "" (để khớp exclude_codes)
col = 'Actual Attendance'
filtered_report_df[col] = (
    filtered_report_df[col]
    .astype(str)
    .str.strip()
    .fillna("")      # nếu muốn NaN coi như rỗng
)

# 3) Lọc bỏ các dòng có mã nằm trong danh sách exclude_codes
filtered_report_df = (
    filtered_report_df.loc[~filtered_report_df[col].isin(exclude_codes)]
    .reset_index(drop=True)
)

#Lọc bỏ những dòng mà Actual Attendance là CO nhưng Ramco là WO

filtered_report_df = filtered_report_df[
    ~((filtered_report_df['Actual Attendance'] == "CO") & 
      (filtered_report_df['Ramco Attendance'] == "WO"))
].reset_index(drop=True)

filtered_report_df
```

```python
#notmark_df.head()

notmark_pivot = pd.pivot_table(
    notmark_df,
    index=['Supervisor Name'],      # hoặc thay bằng cột khác tùy bạn
    columns=['Date'],
    values='Ramco Attendance',
    aggfunc='count',                # hoặc 'sum' nếu là số
    fill_value=0,
    margins=True,                   # thêm Grand Total cho cả hàng và cột
    margins_name='Grand Total'      # tên hiển thị
)

# Hiển thị kết quả
notmark_pivot
```
