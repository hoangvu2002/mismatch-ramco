```python

import pandas as pd
import os

# Hiển thị đủ số dòng và số cột mong muốn
pd.set_option("display.max_rows", 100)   # tối đa số dòng hiển thị
pd.set_option("display.max_columns", None)  # hiện tất cả các cột
pd.set_option("display.width", None)   # không giới hạn độ rộng
```

```python

# --- 1. Đọc dữ liệu từ file master ---
excel_path = r"C:\Users\huuchinh.nguyen\Concentrix Corporation\WFM-Expedia-HCM - Branding files\Headcount\HC Master Database - 2025.xlsx"

# Đọc hai sheet Active và Inactive
df_active = pd.read_excel(excel_path, sheet_name="Active")
df_inactive = pd.read_excel(excel_path, sheet_name="Inactive")

# Kiểm tra kích thước dữ liệu
print("Active:", df_active.shape, "rows, columns")
print("Inactive:", df_inactive.shape, "rows, columns")

# Đường dẫn xuất file
active_csv = "Active.csv"
inactive_csv = "Inactive.csv"

# Xuất ra CSV, ghi đè nếu đã tồn tại
df_active.to_csv(active_csv, index=False)
df_inactive.to_csv(inactive_csv, index=False)

print(f"Đã ghi đè hai file:\n - {os.path.abspath(active_csv)}\n - {os.path.abspath(inactive_csv)}")
df_active.head()
```
